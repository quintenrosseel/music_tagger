# Get All Tags


```python
raw_df.filter(
    (pl.col("MyTagNames").list.len() > 0) &
     (pl.col("MyTagNames").list.contains("TAG YEAR"))
    
)
```

```python
tag_cols = ["MyTagNames", "MyTagIDs"]

df_with_zipped_tags = (
        raw_df
        .select(tag_cols)
        .with_columns(
            pl.struct(tag_cols)
            .map_elements(
                lambda row: list(zip(*[row[c] for c in tag_cols])),
                return_dtype=pl.List(pl.Struct([pl.Field("name", pl.Utf8), pl.Field("id", pl.Utf8)]))
            )
        .alias("TagPairs")
    )
)

exploded_tags = (
    df_with_zipped_tags
    .select(pl.col("TagPairs"))
    .explode("TagPairs")
    .unnest("TagPairs")
)

# Get unique tags
unique_tags = (
    exploded_tags
    .unique()
    .sort("name")
    .filter(pl.col("name").is_not_null())
)

unique_tags
```
